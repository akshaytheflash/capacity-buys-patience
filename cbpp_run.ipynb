```python
import os, subprocess, sys
from google.colab import drive, userdata
drive.mount("/content/drive")
# Token is stored as a Colab Secret named GH_TOKEN (never in the repo)
os.environ["GH_TOKEN"] = userdata.get("GH_TOKEN")
REPO = "akshaytheflash/capacity-buys-patience"
DEST = "/content/capacity-buys-patience"
URL = "https://x-access-token:" + os.environ["GH_TOKEN"] + "@github.com/" + REPO + ".git"
os.path.isdir(DEST) or subprocess.run(["git","clone","-q",URL,DEST], check=True)
os.chdir(DEST)
subprocess.run([sys.executable,"-m","pip","install","-q","-r","requirements.txt"], check=False)
os.environ["CBPP_DRIVE_DIR"] = "/content/drive/MyDrive/cbpp"
os.makedirs(os.environ["CBPP_DRIVE_DIR"], exist_ok=True)
import torch
print("torch", torch.__version__, "| cuda:", torch.cuda.is_available())
print(torch.cuda.get_device_name(0) if torch.cuda.is_available() else "no gpu")
print("repo ready at", DEST)
```
